```python
def fetch_processor(text):
    if text == "Intel Core i7" or text == "Intel Core i5" or text == "Intel Core i3":
        return text
    else:
        if text.split()[0] == "Intel":
            return "Other Intel Processor"
        else:
            return "AMD Processor"
```

# Feature engineering for Memory

```python
df["Memory"].value_counts()
```

```python
df["Memory"] = df["Memory"].astype(str).replace("\.0", "", regex=True)
df["Memory"] = df["Memory"].str.replace("GB", "")
df["Memory"] = df["Memory"].str.replace("TB", "000")
new = df["Memory"].str.split("+", n=1, expand=True)

df["first"] = new[0]
df["first"] = df["first"].str.strip()

df["second"] = new[1]

df["Layer1HDD"] = df["first"].apply(lambda x: 1 if "HDD" in x else 0)
df["Layer1SSD"] = df["first"].apply(lambda x: 1 if "SSD" in x else 0)
df["Layer1Hybrid"] = df["first"].apply(lambda x: 1 if "Hybrid" in x else 0)
df["Layer1Flash_Storage"] = df["first"].apply(
    lambda x: 1 if "Flash Storage" in x else 0
)

df["first"] = df["first"].str.replace(r"\D", "", regex=True)

df["second"].fillna("0", inplace=True)

df["Layer2HDD"] = df["second"].apply(lambda x: 1 if "HDD" in x else 0)
df["Layer2SSD"] = df["second"].apply(lambda x: 1 if "SSD" in x else 0)
df["Layer2Hybrid"] = df["second"].apply(lambda x: 1 if "Hybrid" in x else 0)
df["Layer2Flash_Storage"] = df["second"].apply(
    lambda x: 1 if "Flash Storage" in x else 0
)

df["second"] = df["second"].str.replace(r"\D", "", regex=True)

df["first"] = df["first"].astype(int)
df["second"] = df["second"].astype(int)

df["HDD"] = df["first"] * df["Layer1HDD"] + df["second"] * df["Layer2HDD"]
df["SSD"] = df["first"] * df["Layer1SSD"] + df["second"] * df["Layer2SSD"]
df["Hybrid"] = df["first"] * df["Layer1Hybrid"] + df["second"] * df["Layer2Hybrid"]
df["Flash_Storage"] = (
    df["first"] * df["Layer1Flash_Storage"] + df["second"] * df["Layer2Flash_Storage"]
)

df.drop(
    columns=[
        "first",
        "second",
        "Layer1HDD",
        "Layer1SSD",
        "Layer1Hybrid",
        "Layer1Flash_Storage",
        "Layer2HDD",
        "Layer2SSD",
        "Layer2Hybrid",
        "Layer2Flash_Storage",
    ],
    inplace=True,
)
```

# Model Training

```python
# extracting X and Y
X = df.drop(columns=["Price"])
y = df["Price"]
```

```python
X
```

```python
y
```

```python
# Train-Test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import KFold, cross_validate
from sklearn.metrics import make_scorer, r2_score, mean_absolute_error
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
```

```python
# all models are defined in a dictionary for easy access and iteration.
models = {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(alpha=10),
    "Lasso": Lasso(alpha=0.001),
    "KNN": KNeighborsRegressor(n_neighbors=3),
    "Decision Tree": DecisionTreeRegressor(max_depth=8, random_state=42),
    "SVR": SVR(kernel="rbf", C=10000, epsilon=0.1),
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=3,
        max_samples=0.5,
        max_features=0.75,
        max_depth=15,
    ),
}
```

```python
# categorical column positions.
categorical_columns = [0, 1, 7, 10, 11]
```

```python
# pipeline function to create a pipeline for each model with preprocessing steps.
def make_pipeline(model):
    preprocess = ColumnTransformer(
        transformers=[
            (
                "categorical",
                OneHotEncoder(
                    handle_unknown="ignore", drop="first", sparse_output=False
                ),
                categorical_columns,
            )
        ],
        remainder=StandardScaler(),
    )
    return Pipeline([("preprocess", preprocess), ("model", model)])
```

```python
scoring = {
    "r2": "r2",
    "mae": "neg_mean_absolute_error"
}
```

```python
cv = KFold(n_splits=5, shuffle=True, random_state=42)
results = []

for name, model in models.items():
    scores = cross_validate(
        make_pipeline(model), X_train, y_train, cv=cv, scoring=scoring, n_jobs=-1
    )

    results.append(
        {
            "Model": name,
            "Mean CV R2": scores["test_r2"].mean(),
            "Mean CV MAE (₹)": -scores["test_mae"].mean(),
            "MAE variation (₹)": scores["test_mae"].std(),
        }
    )
```

```python
results_df = pd.DataFrame(results).sort_values("Mean CV MAE (₹)")
results_df
```

```python
best_name = results_df.iloc[0]["Model"]
best_pipe = make_pipeline(models[best_name])
best_pipe.fit(X_train, y_train)

print("Best CV model:", results_df.iloc[0]["Model"])
print("Test R2:", r2_score(y_test, best_pipe.predict(X_test)))
print("Test MAE (₹):", mean_absolute_error(y_test, best_pipe.predict(X_test)))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
colors = ["skyblue"]

plt.barh(
    results_df["Model"],
    results_df["Mean CV MAE (₹)"],
    color=colors
)

plt.xlabel("Mean Cross-Validation MAE (₹)")
plt.ylabel("Model")
plt.title("Model Comparison — Laptop Price Prediction")
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```
